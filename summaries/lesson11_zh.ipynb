好了，我们现在走上正轨了。期望值。然后我们可以开始读关于期望值是什么——你可能会记得，因为我们在高中时学过一点（至少在澳大利亚是）。期望值就是说"这个东西可能的值是多少"。比如说，抛一枚硬币，可能是正面或反面。你想知道它有多少次是正面。我们把正面叫1，反面叫0。抛几次，得到1, 0, 0, 1, 1, 0, 1, 0, 1等等。然后你可以计算平均值——总和除以次数。但期望值是"你期望发生什么"。我们可以通过把所有可能性的概率乘以对应的值然后加起来来计算。

在抛硬币的例子中，两种可能性是正面或反面。正面的概率是0.5，得分是1。反面的概率是0.5，得分是0。所以期望值是0.5×1 + 0.5×0 = 0.5。所以如果正面算赢，抛硬币的期望得分是0.5。

再举一个例子：掷骰子，想知道期望得分是多少。同样，我们可以掷很多次，求和，除以次数，得到平均值。但期望值更一般地是：所有可能性的概率乘以对应的分数的和。掷骰子的可能性是1, 2, 3, 4, 5, 6。每个概率是1/6。得分就是数字本身。所以期望值是1/6 + 2/6 + 3/6 + 4/6 + 5/6 + 6/6 = 3.5。这就是期望值的含义。

最简单的广播：操作标量和张量。a > 0就是a大于[0,0,0]——标量0被广播到整个张量。APL里也一样：a < 5。

举个例子：向量c = [10, 20, 30]，矩阵m = 1-9的3×3矩阵。c和m形状不同（3 vs 3×3），但我们可以把它们加起来。发生了什么？[10,20,30]加到了第1行[1,2,3]，再加到第2行[4,5,6]，再加到第3行[7,8,9]——所以向量被广播到每一行。

背后的机制是expand_as()。t = c.expand_as(m)创建了一个"看起来像"3×3的矩阵，每行都是[10,20,30]，但实际上没有复制数据——在内存中只存储了[10,20,30]，但stride（步幅）为0，行间跳转时位置不变。这种stride技术是NumPy和PyTorch用来高效实现扩展、转置等操作的核心。

c[None]在位置0插入新的单位轴，形状从(3,)变成(1,3)。我从来不用unsqueeze，我更喜欢用None索引——更灵活。c[:, None]在位置1插入单位轴，变成(3,1)。c[None]等同于unsqueeze(0)。因为尾部冒号可选，c[None]就够。...（省略号）表示所有维度，所以c[..., None]总是在末尾加单位轴。

c[:, None]和expand_as(m)会把列向量[10,20,30]复制成3×3矩阵（每列相同）。m + c[:, None]就是把向量加到每一列——[10,20,30]加到各列。

所以m + c[None,:]是把[10,20,30]加到每一行；m + c[:, None]是把[10,20,30]加到每一列。

那c[None,:] × c[:, None]呢？(1×3) × (3×1) ——广播规则：维度为1的被复制以匹配对方。结果得到3×3的外积。不需要特殊函数，只需广播。还可以做外积布尔运算。

广播规则：比较两个张量的形状时，从右往左比较。维度兼容的条件是相等或者其中一个是1。如果某个维度缺失，视为1。所以对于256×256×3的彩色图像和长度为3的一维数组（如RGB缩放因子），从右往左比较：3=3，然后图像有256而向量没有（视为1），所以向量被广播到所有256×256个像素。这就是fastai库中归一化图像的方法。

## DiffEdit 的核心思想

DiffEdit 想解决的问题是：**用文字来编辑图片中的特定物体，而不需要手动画出要修改的区域**。

比如，你有一张"马"的图片，输入"斑马"，它就能自动把马变成斑马，而背景完全不变。

---

## 三个步骤，一个比一个巧妙

### 第一步：自动生成遮罩（最精彩的一步）

这是整个方法最核心的创意。具体做法是：

1. 对原始图片加噪声
2. **两次去噪**：
   - 第一次告诉模型真相：**"这是一匹马"** → 模型估算出"马版本"的噪声
   - 第二次**说谎**：**"这是一匹斑马"** → 模型估算出"斑马版本"的噪声
3. 把两个噪声预测**相减**

关键洞察：当你说"这是斑马"时，模型发现马的那些特征（毛色、条纹分布等）**全部不符合斑马**，于是认为这些区域的像素"都是噪声"。而背景（草地、天空）对斑马来说没问题，所以噪声差异很小。

把差异二值化，就得到了一个**自动生成的遮罩**——它告诉我们应该改哪里，不改哪里。

> Jeremy 特别提到：这个遮罩往往比物体实际区域稍大一点，**这反而是一件好事**，因为边缘细节通常也需要微调。

### 第二步：对原图加噪声

把原始图片加噪声到某个程度，得到一个"带噪声的版本" \(x_r\)。

### 第三步：用遮罩指导的去噪

逐步骤去噪，同时**每个步骤后**把遮罩外的像素从原始图片的噪声版本中复制回来。

这样，只有马（遮罩内区域）被改成了斑马，背景（遮罩外区域）保持不变。

---

## 为什么这个思路很聪明？

Jeremy 强调了这个方法的**直观性**：

- 传统的做法需要**手动画遮罩**——很麻烦
- 直接用扩散模型生成整张图会**改变构图**
- DiffEdit 利用了扩散模型本身对"什么属于这个物体"的知识——模型知道马的哪些像素"不像斑马"，自然就定位出了要修改的区域

---

## 局限（Mikołaj 在聊天中提出的好问题）

这种方法只适用于**不改背景**的编辑。如果你想改整体色调、加散景效果等，它就不适用了，因为遮罩外的像素会被原封不动地保留。

什么是范数

##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

**范数**就是测量"长度"或"大小"的一种方式。

---

### 一维世界（标量）

数轴上一个数离0有多远？我们用**绝对值**：
- |5| = 5，|-3| = 3

**绝对值就是标量的范数。** 所以 Jeremy 和 Siva 说：当你意识到这一点，这是个"敏锐的发现"——原来范数和绝对值是一家人。

---

### 二维/多维世界（向量）

假设你在平面上的位置是 (3, 4)。你离原点 (0, 0) 有多远？

用勾股定理：√(3² + 4²) = 5

这就是向量的 **L2 范数**（最常用的一种），又叫欧几里得范数。公式是：

**\|v\|₂ = √(v₁² + v₂² + ... + vₙ²)**

---

### 再高一点：矩阵

矩阵的 Frobenius 范数就是把这个概念推广到二维：
- 把所有元素平方
- 求和
- 开根号

```
m = [[1, 2],
     [3, 4]]
     
‖m‖_F = √(1² + 2² + 3² + 4²) = √30
```

---

### 那你可能会问："为什么要有这么多不同的范数？"

因为"长度"这个概念在不同的场景下有不同的定义方式：

| 范数 | 公式 | 直观理解 |
|------|------|----------|
| L1（曼哈顿） | \|x₁\| + \|x₂\| + ... | 只能沿网格走，像曼哈顿街道 |
| L2（欧几里得） | √(x₁² + x₂² + ...) | 直线距离，日常理解的"长度" |
| L∞（最大值） | max(\|x₁\|, \|x₂\|, ...) | 只看最大的分量 |

---

Jeremy 在课上用 Frobenius 范数，是因为它**简单好用**——平方、求和、开根号，而 PyTorch 里一句 `(m*m).sum().sqrt()` 就能搞定。
